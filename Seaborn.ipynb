```python
# Data Visualization using Univariate Analysis: 
# 1. Numerical Variables (Data) 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("./IT_Employees_Info.csv"); 

print(df.head());
```

# Univariate Analysis

```python
# Set the overall figure size 
plt.figure(figsize=(30, 10)) 
# 1. Histogram - Age 
plt.subplot(3, 3, 2); 
sns.histplot(df['Salary'], bins=5, kde=True, color='skyblue')
plt.title('Histogram of Age')
plt.xlabel('Salary')
plt.ylabel('Count'); 
plt.show() 
```

```python
# 2. KDE Plot - Salary(Kernel Density Estimation) 
plt.subplot(3, 2, 2)
sns.kdeplot(df['Salary'], fill=True, color='green') 
plt.title('KDE Plot of Salary') 
plt.xlabel('Salary') 
plt.ylabel('Density')  
plt.show() 
```

```python
# 3. Distribution Plot - Experience
plt.subplot(3, 2, 3)
sns.histplot(df['Salary'], kde=True, color="green")
plt.title('Distribution of Experience (Years)')
plt.xlabel('Experience (Years)') 
plt.ylabel('Count')
plt.show() 
```

```python
# 4. Boxplot - Age
plt.subplot(3, 2, 4)
sns.boxplot(x=df['Salary'], color='purple')
plt.title('Boxplot of Salary')
plt.xlabel('Salary')
plt.show() 
```

```python
# 5. Violinplot - Salary
plt.subplot(3, 2, 5)
sns.violinplot(x=df['Salary'], color='green')
plt.title('Violin Plot of Salary')
plt.xlabel('Salary')
plt.show() 
```

# 2. Categorical Variables (Data) 

```python
# Bar Graph - Department
plt.figure(figsize=(5, 5))
df['Department'].value_counts().plot(kind='pie', color='skyblue')
plt.title('Bar Graph - Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.tight_layout()   
plt.show()  
```

```python
# Pie Chart - Department Distribution
plt.figure(figsize=(5, 4))
df['Department'].value_counts().plot.pie(autopct='%1.2f%%',shadow=True); 
plt.title('Pie Chart of Department Distribution')
# Adjust layout
plt.tight_layout()
plt.show();  
```

# Bivariate Analysis

```python
# 1. Numerical vs Numerical
# Graphs: Line Plot, Scatterplot, LMplot, Jointplot, Heatmap
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('Product_Sales_Cat.csv');  
```

```python
# 1️ Line Plot
plt.figure(figsize=(6,4))
sns.lineplot(data=df.sort_values('Price'), x='Price', y='Revenue')
plt.title("Line Plot: Price vs Revenue")
plt.show()

# Insight:
# As Price increases, Revenue also increases — showing a positive trend.
```

```python
# 2️ Scatter Plot
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Price', y='Revenue', size='Quantity', hue='Discount', palette='viridis')
plt.title("Scatter Plot: Price vs Revenue (Discount & Quantity)"); 
plt.show()

# Insight:
# Each point = a product. Bigger size = higher Quantity.
# Color shows Discount — higher discounts might reduce Revenue.
```

```python
# 3️ LM Plot (Regression Line); 
sns.lmplot(data=df, x='Price', y='Revenue', height=5, aspect=1.2)
plt.title("LM Plot: Price vs Revenue (with Regression Line)")
plt.show(); 
#  Insight:
# Regression line shows overall trend — Revenue grows linearly with Price. 
```

```python
# 4️ Joint Plot
sns.jointplot(data=df, x='Quantity', y='Revenue', kind='scatter', color='teal')
plt.suptitle("Joint Plot: Quantity vs Revenue", y=1.02)
plt.show() 

# Insight:
# Helps to see how Revenue changes with Quantity + distribution of both.  
```

### 5️ Heatmap
    A heatmap is used to show the relationship between numerical variables.
    In this chart, Quantity and Price have a strong effect on Revenue, while Discount has very little effect.


```python
plt.figure(figsize=(6,4))
corr = df.select_dtypes(include=['int64', 'float64']).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Heatmap: Numerical Correlations (Product Sales)")
plt.show(); 

# Insight:
# High correlation between Price & Revenue → strong relationship.
# Discount has a negative correlation with Revenue.

# Quantity → Revenue = 0.75

# Strong relationship
# Selling more quantity clearly increases revenue
# Most important insight
```

```python
# Price → Revenue = 0.65
# Good relationship
# Higher price → higher revenue 
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load the updated dataset
df = pd.read_csv('Product_Sales_Cat.csv')

# Boxplot: Revenue by Category 
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Category', y='Revenue')
plt.title("Boxplot: Revenue by Product Category"); 
plt.show();  
```

# Multivariate Analysis 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('Product_Sales_Cat.csv')
```

```python
# Pair Plot
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

sns.pairplot(df[numeric_cols], diag_kind='kde')  # diag_kind='kde' for smooth distributions

plt.suptitle("Pair Plot - Multivariate Relationships", y=1.02); 
plt.show(); 

# Insight:
# Shows relationships between all numerical columns (Price, Quantity, Discount, Revenue).
# Diagonal shows the distribution of each variable.
```

```python
# Heatmap (Correlation Heatmap)
plt.figure(figsize=(10, 6))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f"); 
plt.title("Correlation Heatmap - Product Sales");
plt.show();  
```

```python
# Bubble Chart (Scatter with Size and Hue)
plt.figure(figsize=(10, 6)); 
sns.scatterplot(data=df,
    x='Price',
    y='Revenue',
    size='Quantity',      # bubble size
    hue='Category',       # color by Category
    palette='Set2',
    sizes=(20, 200)       # min and max bubble size
)
plt.title("Bubble Chart: Price vs Revenue (Size=Quantity, Color=Category)")
plt.legend(bbox_to_anchor=(1.05, 1), loc=2)  # move legend outside
plt.show()

# Insight:
# Bigger bubbles = higher quantity sold.
# You can see how different categories perform in terms of Price vs Revenue.  
```

## 3D Scatter Plot  

```python
import matplotlib.pyplot as plt
import pandas as pd
from mpl_toolkits.mplot3d import Axes3D

# Load data
df = pd.read_csv('IT_Employees_Info.csv'); 

# Count how many entries (employees) are in each department
dept_counts = df['Department'].value_counts(); 
print("Number of employees in each department:\n")
print(dept_counts);  
```

```python
# Prepare figure
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Set unique color for each department
colors = ['red', 'blue', 'green', 'orange', 'purple', 'cyan', 'brown', 'magenta', 'yellow']
departments = df['Department'].unique()

# Plot each department separately
for i, dept in enumerate(departments):
    data = df[df['Department'] == dept]
    ax.scatter(
        data['Experience(Years)'],
        data['Age'],
        data['Salary'],
        color=colors[i % len(colors)],
        label=f"{dept} ({len(data)})",  # Show count in legend
        s=50
    )

# Labels and title
ax.set_xlabel('Experience (Years)')
ax.set_ylabel('Age')
ax.set_zlabel('Salary')
ax.set_title('3D Scatter Plot of Employees by Department')
ax.legend(title='Department (Count)'); 
plt.tight_layout()
plt.show(); 

#  Sequential colormaps (good for data that goes from low to high):
# 'viridis' (default, great for most uses)

# 'plasma'
# 'inferno'
# 'magma'
# 'cividis'
# 'Greens'
# 'Blues'
# Diverging colormaps (for data with a meaningful midpoint, like +/−):
# 'coolwarm'
# 'bwr' (blue-white-red)
# 'seismic'
# 'PiYG'
#  Categorical/discrete colormaps (for categories — less common in scatter):
# 'tab10'
# 'Set1'
# 'Accent'
```

```python
# 5. Stacked Bar Chart (e.g., Gender-wise count per Department)
# Group data
stacked_data = df.groupby(['Department', 'Gender']).size()

# Plot
stacked_data.plot(kind='bar', figsize=(10, 6), colormap='viridis')
plt.title('Stacked Bar Chart: Gender Distribution per Department')
plt.ylabel('Number of Employees') 
plt.xticks(rotation=50) 
plt.tight_layout() 
plt.show()  
```
