# String methods

```python
# PYTHON STRING METHODS
# 1. upper()
text1 = "hello"; 

print("Upper :" , text1.upper()); 
```

```python
# 2. lower()
text2 = "HELLO"; 
print("2. lower()       :", text2.lower())

# 4. title()
text4 = "hello world"
print("4. title()       :", text4.title())

# 5. swapcase() 
text5 = "HeLLo"
print("5. swapcase()    :", text5.swapcase())

# 6. split() IMP 
text6 = "a b c"
print("6. split()       :", text6.split())

# 7. replace()
text7 = "apple"
print("7. replace()     :", text7.replace("a", "A"))

# 8. strip() IMP 
text8 = "   hello   "
print("8. strip()       :", text8.strip())

# 9. startswith()
text9 = "python"
print("9. startswith()  :", text9.startswith("py"))

# 10. endswith()
text10 = "python"
print("10. endswith()   :", text10.endswith("on"))

# 11. isalpha()  
text11a = "abc"
text11b = "abc123"
print("11. isalpha()    :", text11a.isalpha()); 
print("11b. isalpha()   :", text11b.isalpha())

# 12. isnumeric()
text12a = "123"
text12b = "12a"
print("12. isnumeric()  :", text12a.isnumeric())
print("12b. isnumeric() :", text12b.isnumeric())

# 13. isalnum()
text13a = "abc123"
text13b = "abc 123"
print("13. isalnum()    :", text13a.isalnum())
print("13b. isalnum()   :", text13b.isalnum())

# 14. islower()
text14 = "hello"
print("14. islower()    :", text14.islower())

# 15. isupper() 
text15 = "HELLO"
print("15. isupper()    :", text15.isupper())

# 16. find()
text16 = "apple"
print("16. find('p')    :", text16.find("p"))
print("16b. find('z')   :", text16.find("z"))

# 17. join()
words = ["abc", "def", "ghi"]
space_join = " ".join(words)
dash_join = "-".join(words)
comma_join = ",".join(words)

print("17a. join(space) :", space_join)
print("17b. join(-)     :", dash_join)
print("17c. join(,)     :", comma_join);  
```

# MINI REAL-WORLD DATA PROCESSING

```python
print("\n========== MINI DATA PROCESSOR ==========\n")

name = input("Enter your name: ").strip(); 
age = input("Enter your age: ").strip(); 
email = input("Enter your email: ").strip(); 

print("\n---------------- VALIDATION ----------------")

# Name validation
if name.isalpha():
    print("Name  : Valid")
else:
    print("Name  : Invalid (only letters allowed)")

# Age validation
if age.isnumeric():
    print("Age   : Valid")
else:
    print("Age   : Invalid (only numbers allowed)")

# Email validation
if "@" in email and "." in email:
    print("Email : Valid")
else:
    print("Email : Invalid"); 
```

# DATA CLEANING FOR ANALYSI

```python
raw_data = [
    "  pytHon  ",
    "DaTa ScIenCe",
    " MACHINE leArninG  "
    "  deep learning  "
];  

cleaned_data = [i.strip().title() for i in raw_data]; 

print("Raw Data     :", raw_data);

print("\n \nCleaned Data :", cleaned_data);  
```

```python
# Data Parsing (CSV)
print("=== CSV Parsing ===")

csv_row = "          AMan         ,  25   ,  daTa ANalysT   ,PytHon,sqL";

fields = [field.strip().capitalize() for field in csv_row.split(",")];  # Split and clean each field

print("CSV Row:", csv_row);

print("---------------------------------------------------------------------------");

print("Parsed Fields:", fields);  
```
