```python
def calculate(num1,operator,num2):
    if operator=='+':
        return num1 + num2
    elif operator=='-':
        return num1 - num2
    elif operator=='*':
        return num1 * num2
    elif operator=='/':
        return num1 / num2
    else:
        return "Invalide operator"

num1=eval(input("Enter a num"))
operator=input("Select a operator(+,-,*,/:)")
num2=eval(input("enter a another num"));

print(calculate(num1,operator,num2)); 
```

# map()

```python
numbers = [1, 2, 3, 4]; 
squared = list(map(lambda x: x - 2, numbers))
print(squared);  
```

# filter() 

```python
numbers = [1, 2, 3, 4, 5, 6];  
numbers = list(filter(lambda x: x < 3, numbers))
print(numbers);   
```
